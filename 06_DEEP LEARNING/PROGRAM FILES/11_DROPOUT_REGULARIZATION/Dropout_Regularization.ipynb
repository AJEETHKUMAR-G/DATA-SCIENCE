## LIBRARY IMPORT

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
import warnings
warnings.filterwarnings('ignore')
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix,classification_report
```

## DATA IMPORT

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/11_DROPOUT_REGULARIZATION/sonar_dataset.csv',header = None)
```

```python
x = df.drop(60,axis = 1)
y = df[60]
```

```python
y
```

## ONE HOT ENCODING

```python
y = pd.get_dummies(y,drop_first=True).astype(int)
y.sample(5) # R = 1, M = 0
```

```python
y.value_counts()
```

## TRAIN TEST SPLIT

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state =1 )
```

```python
x_train.shape
```

```python
x_test.shape
```

## NN MODEL

```python
model = keras.Sequential([
    keras.layers.Dense(60,input_dim = 60,activation = 'relu'),
    keras.layers.Dense(30,activation = 'relu'),
    keras.layers.Dense(15,activation = 'relu'),
    keras.layers.Dense(1,activation = 'sigmoid')
])
model.compile(
    optimizer ='adam',
    loss = 'binary_crossentropy',
    metrics = ['accuracy']
)
model.fit(x_train,y_train,epochs = 100,batch_size = 8)
```

```python
model.evaluate(x_test,y_test)
```

```python
model.predict(x_test).reshape(-1)
```

```python
y_pred = model.predict(x_test).reshape(-1)
print(y_pred[:10])
```

```python
y_pred = np.round(y_pred)
print(y_pred[:10])
```

```python
y_test[:10]
```

```python
print(classification_report(y_test,y_pred))
```

```python
model_drop = keras.Sequential([
    keras.layers.Dense(60,input_dim = 60,activation = 'relu'),
    keras.layers.Dropout(0.5), # 50%
    keras.layers.Dense(30,activation = 'relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(15,activation = 'relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(1,activation = 'sigmoid')
])
model_drop.compile(
    optimizer ='adam',
    loss = 'binary_crossentropy',
    metrics = ['accuracy']
)
model_drop.fit(x_train,y_train,epochs = 100,batch_size = 8)
```

```python
model_drop.evaluate(x_test,y_test)
```

THE ACCURACY OF TEST SET AFTER USING DROPOUT IS HIGHER

```python
y_pred_drop = model_drop.predict(x_test).reshape(-1)
y_pred_drop = np.round(y_pred_drop)
print(classification_report(y_test,y_pred_drop))
```
