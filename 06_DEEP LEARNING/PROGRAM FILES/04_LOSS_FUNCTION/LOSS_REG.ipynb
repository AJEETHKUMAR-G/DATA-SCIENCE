```python
from tensorflow import keras
from tensorflow.keras import layers
import tensorflow as tf
import numpy as np
(X_train, y_train), (X_test, y_test) = keras.datasets.boston_housing.load_data()
```

```python
X_train.shape
```

```python
model = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)
])
```

```python
model.compile(
    optimizer = 'adam',
    loss = 'mse', # 'mae',keras.losses.Huber(delta=1.0)
    metrics = ['mae']
)
```

```python
model.fit(X_train,y_train,epochs = 50)
```

```python
model.evaluate(X_test,y_test)
```

```python
model1 = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)
])
```

```python
def squared_error(y_true, y_pred):
    return tf.math.square(y_true - y_pred)

def absolute_error(y_true, y_pred):
    return tf.abs(y_true - y_pred)

def huber_loss(y_true, y_pred, delta=1.0):
    error = y_true - y_pred
    condition = tf.abs(error) <= delta
    
    small_error = 0.5 * tf.square(error)
    large_error = delta * (tf.abs(error) - 0.5 * delta)
    
    return tf.where(condition, small_error, large_error)


model1.compile(
    optimizer='adam',
    loss=squared_error, # absolute_error
    metrics = ['mae']
)
```

```python
model1.fit(X_train,y_train,epochs = 50)
```

```python
model1.evaluate(X_test,y_test)
```
