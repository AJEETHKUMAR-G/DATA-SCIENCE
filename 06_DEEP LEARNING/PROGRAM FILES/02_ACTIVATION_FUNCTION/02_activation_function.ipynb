```python
import matplotlib.pyplot as plt
import math
```

## RELU

```python
def relu(x):
    return max(0.0,x)
ip = [x for x in range(-100,100)]
op = [relu(x) for x in ip]
plt.plot(ip,op)
plt.show()
```

## SIGMOID



```python
def sigmoid(x):
    return 1/(1+ math.exp(-x))
ip = [x for x in range(-10,10)]
op = [sigmoid(x) for x in ip]
plt.plot(ip,op)
plt.show()
```

```python
def tanh(x):
    return (math.exp(x) - math.exp(-x))/ (math.exp(x) + math.exp(-x))
ip = [x for x in range(-10,10)]
op = [tanh(x) for x in ip]
plt.plot(ip,op)
plt.show()
```

## LINEAR ACTIVATION 

```python
def linear(x):
    return x
ip = [x for x in range(-10,10)]
op = [linear(x) for x in ip]
plt.plot(ip,op)
plt.show()
```

## SOFTMAX

```python
ip = [11.0, 12.0, 5.0, 2.0]
def softmax(arr):
    exp_vals = [math.exp(x) for x in arr]
    total = sum(exp_vals)
    return [val / total for val in exp_vals]

op = softmax(ip)
op
```

```python
def leaky_relu(x):
    return max(0.1*x,x)
```
