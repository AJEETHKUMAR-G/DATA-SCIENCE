## LOGISTIC REGRESSION

```python
def log_reg(x_train,x_test,y_train,y_test,weights):
    if weights == -1:
        model = LogisticRegression()
    else:
        model = LogisticRegression(class_weight={0:weights[0],1:weights[1]})
    model.fit(x_train,y_train)
    acc = model.score(x_test,y_test)
    print('Accuracy:',acc,'\n')
    y_pred = model.predict(x_test)
    class_repo = classification_report(y_test,y_pred)
    print(class_repo)
```

```python
weights = -1
log_reg(x_train,x_test,y_train,y_test,weights)
```

```python
weights = [1,1.5]
log_reg(x_train,x_test,y_train,y_test,weights)
```

```python
df2.Churn.value_counts()
```

```python
x = df2.drop('Churn',axis = 1)
y = df2.Churn
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state =15,test_size = 0.2,stratify = y)
```

```python
y_train.value_counts()
```

model1 --> class1(1495) + class0(0, 1495)

model2 --> class1(1495) + class0(1496, 2990)

model3 --> class1(1495) + class0(2990, 4130)

```python
model = LogisticRegression()
df3 = x_train.copy()
df3['Churn'] = y_train
```

```python
df3.sample(3)
```

```python
df_0 = df3[df3.Churn == 0]
df_1 = df3[df3.Churn == 1]
```

```python
def get_train_batch(df_major,df_minor,start,end):
    df_train = pd.concat([df_major[start:end],df_minor],axis = 0)
    x_train = df_train.drop('Churn',axis = 1)
    y_train = df_train.Churn
    return x_train,y_train
```

```python
x_train,y_train = get_train_batch(df_0,df_1,0,1495)
```

```python
model1 = LogisticRegression()
model1.fit(x_train, y_train)
y_pred1 = model1.predict(x_test)
```

```python
x_train,y_train = get_train_batch(df_0,df_1,1495,2990)
model2 = LogisticRegression()
model2.fit(x_train, y_train)
y_pred2 = model1.predict(x_test)
```

```python
x_train,y_train = get_train_batch(df_0,df_1,2990,4130)
model3 = LogisticRegression()
model3.fit(x_train, y_train)
y_pred3 = model1.predict(x_test)
```

```python
y_pred_final = y_pred1.copy()
for i in range(len(y_pred1)):
    n_ones = y_pred1[i] + y_pred2[i] + y_pred3[i]
    if n_ones>1:
        y_pred_final[i] = 1
    else:
        y_pred_final[i] = 0
```

```python
print(classification_report(y_test,y_pred_final))
```
