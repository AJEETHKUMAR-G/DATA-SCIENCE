## DATA WAREHOUSING

```sql
CREATE DATABASE salesWarehouse;
```

### STAGING LAYER

```sql
CREATE TABLE saleswarehouse.stagingDWH LIKE sales.Orders;
```

```sql
SELECT * FROM saleswarehouse.stagingDWH;
```

```sql
INSERT INTO saleswarehouse.stagingdwh SELECT * FROM sales.orders;
```

```sql
SELECT * FROM saleswarehouse.stagingdwh;
```

**OR CAN ALSO WRITE SIMILAR QUERIES : CREATE TABLE salesWarehouse.stagingdwh AS SELECT * FROM sales.Orders**

### APPLYING TRANSFORMATIONS

```sql
SELECT * FROM saleswarehouse.stagingdwh WHERE Quantity IS NOT NULL;
```

**NOW WE HAVE TWO OPTIONS LIKE WE CAN DIRECTLY SHARE THE DATA TO OUR CORE LAYER OR WE CAN CREATE A TABLE OR VIEW(CREATING ANOTHER LAYER)**

```sql
-- HERE WE CREATE A VIEW
CREATE VIEW saleswarehouse.transform_sales AS SELECT * FROM saleswarehouse.stagingdwh WHERE Quantity IS NOT NULL;
```

```sql
SELECT * FROM saleswarehouse.transform_sales;
```

### CORE LAYER

```sql
CREATE TABLE saleswarehouse.core_sales AS
SELECT * FROM saleswarehouse.transform_sales;
```

```sql
SELECT * FROM saleswarehouse.core_sales;
```
