```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
orders = [
    (1001, 101, "Laptop", 55000, 1),
    (1002, 102, "Mobile", 25000, 2),
    (1003, 103, "Headphones", 3000, 3),
    (1004, 104, "Keyboard", 1500, 5),
    (1005, 105, "Mouse", 800, None),
    (1006, 106, "Monitor", 12000, 2),
    (1007, 107, "Printer", 15000, 6)
]

cols = ["order_id","customer_id","product","amount","warehouse_id"]

orders_df = spark.createDataFrame(
    orders,cols
)
```

```python
orders_df.display()
```

```python
warehouses = [
    (1,"Chennai"),
    (2,"Bangalore"),
    (3,"Hyderabad"),
    (4,"Mumbai")
]

warehouse_df = spark.createDataFrame(
    warehouses,
    ["warehouse_id","warehouse_city"]
)

warehouse_df.display()
```

## Inner join

```python
orders_df.join(
    warehouse_df,
    on="warehouse_id",
    how="inner"
).display()
```

## Left join

```python
orders_df.join(
    warehouse_df,
    on="warehouse_id",
    how="left"
).display()
```

## Right join

```python
orders_df.join(
    warehouse_df,
    on="warehouse_id",
    how="right"
).display()
```

## Outer join

```python
orders_df.join(
    warehouse_df,
    on="warehouse_id",
    how="outer"
).display()
```

## Left Anti join

"Show me all rows from the LEFT table that do NOT have a matching row in the RIGHT table."

```python
orders_df.join(
    warehouse_df,
    on="warehouse_id",
    how="left_anti"
).display()
```

## Left semi joins

Left Semi Join returns only the rows from the LEFT DataFrame that have a matching row in the RIGHT DataFrame.

```python
orders_df.join(
    warehouse_df,
    on="warehouse_id",
    how="left_semi"
).display()
```
