```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

Window functions perform calculations across a group of rows while keeping each row intact. Unlike groupBy(), which reduces multiple rows into one, window functions return a value for every row.

```python
data = [
    (101,"Ajeeth","IT",75000),
    (102,"Priya","IT",85000),
    (103,"Rahul","IT",70000),
    (104,"Sneha","HR",60000),
    (105,"Kiran","HR",65000),
    (106,"Anu","HR",60000),
    (107,"Vijay","Finance",90000),
    (108,"Meena","Finance",95000),
    (109,"Arun","Finance",90000),
    (110,"John","IT",85000)
]

columns = ["Emp_ID","Employee","Department","Salary"]

df = spark.createDataFrame(data, columns)
```

```python
df.display()
```

```python
window = Window.partitionBy("Department").orderBy(desc("Salary"))

df.withColumn("Row_Number",
              row_number().over(window)
             ).display()
```

```python
df.withColumn(
    "Rank",
    rank().over(window)
).display()
```

```python
df.withColumn(
    "Dense_Rank",
    dense_rank().over(window)
).display()
```

```python
window = Window.partitionBy("Department").orderBy("Salary")

df.withColumn(
    "Previous_Salary",
    lag("Salary",1).over(window)
).display()
```

```python
df.withColumn(
    "Next_Salary",
    lead("Salary").over(window)
).display()
```

```python
window = Window.partitionBy("Department").orderBy("Salary")

df.withColumn(
    "Running_Total",
    sum("Salary").over(window)
).display()
```

```python
df.withColumn(
    "Running_Avg",
    avg("Salary").over(window)
).display()
```
