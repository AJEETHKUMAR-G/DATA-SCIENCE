```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.read.option('InferSchema',True).option('header',True).csv('/Volumes/distributed_processing/default/py_spark/BigMart Sales.csv')
```

```python
df.printSchema()
```

```python
df_gp = df.groupby('Item_Fat_Content')
```

```python
type(df_gp)
```

```python
df_gp.count().display()
```

```python
df_gp.agg(count("item_type").alias('Count_of_Item_Type')).display()
```

```python
df.groupby('Item_Fat_Content','item_type').count().display()
```

```python
df_gp.sum('Item_MRP').alias('sum_of_price').display()
```

```python
df_gp.sum('Item_MRP').withColumnRenamed("sum(Item_MRP)","Total_Price").display()
```

```python
df_gp.avg('Item_MRP').withColumnRenamed('avg(Item_MRP)',"Average_of_price").\
withColumn("Average_of_price", round("Average_of_price", 2)).display()
```

```python
df_gp.max('Item_MRP').display() # min()
```

```python
df_gp.agg(
    min('Item_MRP').alias('minimum')
).display()
```

```python
df_gp.agg(
    sum("Item_MRP").alias("Total Price"),
    min("Item_Outlet_Sales").alias("Min Outlet Sales"),
    avg("Item_Visibility").alias("Avg Visibility")
).withColumn(
    "Avg Visibility",
    round("Avg Visibility", 2)
).display()
```
