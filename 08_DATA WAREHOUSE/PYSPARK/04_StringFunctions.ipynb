```python
df = spark.createDataFrame(data,columns)
```

```python
df.display()
```

```python
df.select(initcap(col('full_name')).alias('Capitalized')).display() # HERE WE CAN SEE SOME LEADING AND TRAILING SPACES
# TRY lower()/upper()
```

```python
df.select(length(col('phone')).alias('length_of_phn')).display()
```

```python
df.select(
    "full_name",
    trim("full_name").alias("Trimmed")
).display()
```

```python
df.select(
    concat("full_name","job_role").alias("Combined")
).display()
```

```python
df.select(
    concat_ws(" | ","full_name","job_role").alias("Employee")
).display()
```

```python
df.select(
    substring("job_role",1,4).alias("Sub")
).display()
```

```python
df.filter(
    col("job_role").startswith("Data") # TRY endswith()
).display()
```

```python
df.filter(
    col("email").like("%gmail.com")
).display()
```

```python
df.filter(
    col("job_role").ilike("%engineer%")# case insensitive  
).display()
```

```python
df.select(
    "job_role",
    reverse("job_role").alias("Reverse")
).display()
```

```python
df.display()
```

```python
df.select(
    overlay(col("job_role"),lit("Senior "),lit(1)).alias("Updated")
).display()
```

```python
df.select(
    format_string("%s works as %s", col("full_name"), col("job_role")).alias("Description")
).display()
```

```python
df.select(
    "phone",
    regexp_replace("phone","[^0-9]","").alias("CleanPhone")
).display()
```

```python
df.select(
    "email",
    regexp_extract("email","(.*)@",1).alias("Username")
).display()
```
