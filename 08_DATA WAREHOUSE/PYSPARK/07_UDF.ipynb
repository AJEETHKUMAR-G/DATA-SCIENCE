```python
def age_category(x):
    if x > 18:
        return 'Major'
    else:
        return 'Minor'
```

```python
age_detector = udf(age_category,StringType())
```

```python
df.withColumn('age_section',age_detector('age')).display()
```

```python
def salary_grade(salary):

    if salary is None:
        return "Not Available"

    elif salary >= 80000:
        return "High"

    elif salary >= 50000:
        return "Medium"

    else:
        return "Low"

grade_udf = udf(salary_grade, StringType())

df.withColumn("Salary Grade", grade_udf("salary")).display()
```

```python
def create_email(name):

    if name is None:
        return None

    return name.lower().replace(" ",".") + "@company.com"

email_udf = udf(create_email, StringType())

df.withColumn("Email", email_udf("name")).display()
```

```python
def net_salary(salary, bonus):

    if salary is None:
        return None

    if bonus is None:
        bonus = 0

    return salary + bonus

net_udf = udf(net_salary, DoubleType())

df.withColumn("Net Salary", net_udf("salary","bonus")).display()
```

```python
def replace_null_salary(salary):

    if salary is None:
        return 0

    return salary
```

```python
salary_udf = udf(replace_null_salary, IntegerType())

df.withColumn("Updated Salary", salary_udf("salary")).display()
```

```python
def total_income(salary, bonus):

    if salary is None:
        salary = 0

    if bonus is None:
        bonus = 0

    return salary + bonus
```

```python
income_udf = udf(total_income, DoubleType())

df.withColumn(
    "Total Income",
    income_udf("salary", "bonus")
).display()
```
