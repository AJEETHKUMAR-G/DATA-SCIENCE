The MERGE statement is used to:  
INSERT, UPDATE, and DELETE data in a target table based on conditions with a source table.  
SYNTAX FOR MERGE:  
**MERGE target_table AS T  
USING source_table AS S  
ON matching_condition  

WHEN MATCHED THEN  
    UPDATE SET ...  

WHEN NOT MATCHED THEN  
    INSERT (...)**


```sql
MERGE INTO sales_scd.DimProducts AS target
USING sales_scd.viewdimproducts AS source
ON target.productid = source.productid
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *; 
```
